# Exportação de Dados

```python

from pathlib import Path

pasta_exportacao = Path("../tabelas")
pasta_exportacao.mkdir(parents=True, exist_ok=True)

catalog = "vcredit"
schema = "gold"

tabelas = (
    spark.sql(f"SHOW TABLES IN {catalog}.{schema}")
    .select("tableName")
    .collect()
)

total = 0

for linha in tabelas:
    nome = linha["tableName"]

    if not nome.startswith(("dim_", "fact_", "vw_", "resumo_fluxo_clientes")):
        continue
    df = spark.table(f"{catalog}.{schema}.{nome}")
    caminho_csv = pasta_exportacao / f"{nome}.csv"
    (
        df.toPandas()
        .to_csv(
            str(caminho_csv),
            index=False,
            encoding="utf-8-sig"
        )
    )
```
